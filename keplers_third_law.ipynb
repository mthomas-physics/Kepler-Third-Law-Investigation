Investigating Keplers Third Law Using Orbital Simulations

Introduction

Understanding orbital mechanics is essential to furthering space exploration and uncovering more about planet formation. This project will investigate how a planets orbital radius relates to its orbital period while simultaniously testing Keplers third law.

Research Question

How does the orbital radius from the sun affect a planets orbital period?

Hypothesis

I expect for this investigation to show that planets that have a larger orbital radius will have a larger orbital period. In order for a planet with a larger orbital radius to have a smaller period it would need to have a larger velocity. However since the force of gravity is inversely proportional to the orbital radius the required orbital velocity would be smaller. Therefore, I expect that planets with a larger orbital radius will also have a larger orbital period.

Background

Kepler was a mathematician who studied planetary motion and attampted to use math to explain how the planets moved. Kepler developed three laws in his lifetime the third of which states that the square of a planets period is proportionally related to the cube of the planets orbital radius. Keplers Third law allows us to predict the orbital period of a planet or the orbital radius if given at least one. In this investigation of Keplers Third law a computer simulation will be used to vary orbital radii and uncover its resulting periods.

```python
##Method
import matplotlib.pyplot as plt

G = 6.67430e-11  # gravitational constant in m^3 kg^-1 s^-2
M_sun = 1.989e30  # mass of the sun in kg
Time_step = 60 * 60 # time step in seconds (1 hour)
AU = 1.496e11  # astronomical unit in meters

periods = []  # list to store the orbital periods for each radius

radius_values_in_AU = [0.387, 0.500, 0.723, 1.000, 1.500, 1.524, 2.000, 2.500, 3.000]

radius_values_in_meters = [] # list to store the radius values in meters

for radius in radius_values_in_AU:
    radius_values_in_meters.append(radius * AU)

for r in radius_values_in_meters:
    # Calculate the orbital velocity for a circular orbit
    orbital_velocity = (G * M_sun / r) ** 0.5  # in m/s

    x_position = r  # initial x position in meters
    y_position = 0  # initial y position in meters

    x_velocity = 0  # initial x velocity in m/s
    y_velocity = orbital_velocity  # initial y velocity in m/s

    elapsed_time = 0  # elapsed time in seconds

    completed_orbit = False # a condition to check if the orbit is completed

    while not completed_orbit:
        distance = (x_position**2 + y_position**2)**0.5  # distance from the sun in meters

        ax = -G * M_sun * x_position / (distance)**3
        ay = -G * M_sun * y_position / (distance)**3

        x_velocity += ax * Time_step
        y_velocity += ay * Time_step

        y_previous = y_position

        x_position += x_velocity * Time_step
        y_position += y_velocity * Time_step

        elapsed_time += Time_step

        if y_previous < 0 and y_position >= 0 and x_position > 0: # if the planet crosses the positive x-axis the orbit is completed
            completed_orbit = True 

    period = elapsed_time
    periods.append(period)

print("Orbital Periods (seconds):", periods)
print("Radius (meters):", radius_values_in_meters)


plt.plot(radius_values_in_meters, periods, marker='o')
plt.title("Orbital Period (s) vs. Radius (m)")
plt.xlabel("Radius (m)")
plt.ylabel("Orbital Period (s)")
plt.show()

periods_squared = []  # list to store the squares of the orbital periods
radius_cubed = []  # list to store the cubes of the radius values

for P in periods:
    P_squared = P**2
    periods_squared.append(P_squared)
    
for R in radius_values_in_meters:
    R_cubed = R**3
    radius_cubed.append(R_cubed)

print("Orbital Periods Squared (seconds^2):", periods_squared)
print("Radius Values Cubed (meters^3):", radius_cubed)  # convert meters to AU


plt.plot(radius_cubed, periods_squared, marker='o')
plt.title("Orbital Period (s)**2 vs. Radius (m)**3")
plt.xlabel("Radius (m)**3")
plt.ylabel("Orbital Period (s)**2")
plt.show()

```

Results

This simulation ran nine different radii ranging from 0.387 AU to 3.000 AU. The orbital period increased as the radii increased. The relationship was further investigated by plotting the period squared and the radius cubed. The resulting plot showed a linear relationship. 

Disscussion

The results of this simulation supported my hypothesis that larger radii will produce a larger orbital period. The linear realtionship between the sqared period and cubed radius provides evidence for Keplers Third law. The reason for this result is due to the gravitational force decreasing with distance. The larger the radius the smaller the gravitational force meaning a slower velocity and a larger orbit to travel. There were several factors that affected the accuracy of the data. These factors include using the timestep, the assumption of having a perfectly circular orbit, having a stationary sun, and ignoring the other planets.  

Conclusion

This simulation investiaged how the orbital radius of a planet afffects its period. The results showed a proportional relationship between the period squared and the radius cubed. The results also showed the period increasing as the radius increased. The hypothesis of a planets orbital period increasing with its orbital radius was shown to be correct. This investigation supports Kepler's Third law of the period squared being directly proportional to the radius cubed.
